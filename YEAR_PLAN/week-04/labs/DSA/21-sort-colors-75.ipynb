```python
class Solution:
    def sortColors(self, nums):
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        low = mid = 0
        high = n-1
        
        while(mid<=high):
            print(nums)
            if nums[mid] ==0:
                nums[mid],nums[low] = nums[low],nums[mid]
                mid+=1
                low+=1
            elif nums[mid] == 1:
                mid+=1
            elif nums[mid] == 2:
                nums[mid],nums[high] = nums[high],nums[mid]
                high-=1
```

```python
ob = Solution()
nums = [2,0,1]
nums = [2,0,2,1,1,0]
nums = [2,2,1,1,1,2,0]
ob.sortColors(nums)
print(nums)
```
