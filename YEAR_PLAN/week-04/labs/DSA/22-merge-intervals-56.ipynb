```python
class Solution:
    def merge(self, intervals):
        intervals.sort()
        res = [intervals[0]]
        for n in range(1,len(intervals)):
            if res[-1][1] >= intervals[n][0]:
                res[-1][1] = max(intervals[n][1],res[-1][1])
            else:
                res.append(intervals[n])
        return res
```

```python
def test(intervals, expected):
    got = Solution().merge(intervals)
    assert got == expected, f"mismatch: got {got}, expected {expected}"
    print(f"OK  intervals={intervals}  ->  {got}")


# 1. basic overlapping
test([[1,3],[2,6],[8,10],[15,18]],
     [[1,6],[8,10],[15,18]])

# 2. touching intervals (IMPORTANT edge)
test([[1,4],[4,5]],
     [[1,5]])

# 3. no overlap at all
test([[1,2],[3,4],[5,6]],
     [[1,2],[3,4],[5,6]])

# 4. completely nested intervals
test([[1,10],[2,3],[4,8]],
     [[1,10]])

# 5. single interval
test([[5,7]],
     [[5,7]])

# 6. all intervals same
test([[2,5],[2,5],[2,5]],
     [[2,5]])

# 7. unsorted input (VERY IMPORTANT)
test([[8,10],[1,3],[2,6],[15,18]],
     [[1,6],[8,10],[15,18]])

# 8. negative numbers
test([[-10,-1],[-5,0],[1,3]],
     [[-10,0],[1,3]])

# 9. large overlap chain
test([[1,4],[2,5],[3,6]],
     [[1,6]])

# 10. intervals inside each other mixed
test([[1,4],[0,2],[3,5]],
     [[0,5]])

# 11. disjoint + one merge
test([[1,4],[5,6],[2,3]],
     [[1,4],[5,6]])

# 12. touching at multiple points
test([[1,2],[2,3],[3,4]],
     [[1,4]])

# 13. zero-length intervals
test([[1,1],[2,2],[1,2]],
     [[1,2]])

# 14. wide + small intervals inside
test([[1,100],[10,20],[30,40],[50,60]],
     [[1,100]])

# 15. mixed chaos case (real test)
test([[6,8],[1,9],[2,4],[4,7]],
     [[1,9]])

test([[2,3],[2,2],[3,3],[1,3],[5,7],[2,2],[4,6]],
     [[1,3],[4,7]])
```
