# LC #448 — Find All Numbers Disappeared in an Array

Pattern: **Cyclic Sort**

Given an array `nums` of `n` integers where `nums[i]` is in the range `[1, n]`, return an array of all the integers in the range `[1, n]` that do **not** appear in `nums`.

Constraint: try to do it in O(n) time and O(1) extra space (the output array does not count).

```python
class Solution:
    def findDisappearedNumbers(self, nums):
        # cyclic sort: put each number n at index n-1
        res = []
        i=0
        while i < len(nums):
            correct_index = nums[i] - 1
            if(nums[i]!= nums[correct_index]):
                nums[i],nums[correct_index] = nums[correct_index], nums[i]
            else:
                i+=1
        for i in range(len(nums)):
            if nums[i] != i+1:
                res.append(i+1)
        return res
```

```python
ob = Solution()
print(ob.findDisappearedNumbers([2,2]))
```

```python
ob = Solution()

# Test 1: standard case
# expected: [5, 6]
print(ob.findDisappearedNumbers([4,3,2,7,8,2,3,1]))

# Test 2: small case with one duplicate
# expected: [2]
print(ob.findDisappearedNumbers([1,1]))

# Test 3: nothing missing
# expected: []
print(ob.findDisappearedNumbers([1,2,3,4]))

# Test 4: all same number
# expected: [1, 3, 4]
print(ob.findDisappearedNumbers([2,2,2,2]))

# Test 5: single element present
# expected: []
print(ob.findDisappearedNumbers([1]))

# Test 6: reversed order, no missing
# expected: []
print(ob.findDisappearedNumbers([5,4,3,2,1]))
```
