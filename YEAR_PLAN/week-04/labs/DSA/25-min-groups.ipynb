```python
class Solution:
    def minGroups(self, intervals):
        left,right = [],[]
        res = 0        
        
        for l,r in intervals:
            left.append(l)
            right.append(r)
            
        left.sort()
        right.sort()
        
        i=j=0
        while i<len(intervals):
            if(left[i]<=right[j]):
                i+=1
            else:
                j+=1        
            res = max(res,i-j)
        return res
```

```python
ob = Solution()
print(ob.minGroups([[5,10],[6,8],[1,5],[2,3],[1,10]]))
```
