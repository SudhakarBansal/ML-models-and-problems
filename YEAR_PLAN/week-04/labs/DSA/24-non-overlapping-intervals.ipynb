# LC #435 — Non-overlapping Intervals

**Date:** 2026-05-07 (Thu)  
**Pattern:** Intervals  
**Difficulty:** Medium  

## Problem

Given an array of intervals `intervals` where `intervals[i] = [start_i, end_i]`, return the **minimum number of intervals you need to remove** to make the rest of the intervals non-overlapping.

## Constraints
- `1 <= intervals.length <= 10^5`
- `intervals[i].length == 2`
- `-5 * 10^4 <= start_i < end_i <= 5 * 10^4`

## Pattern hint

<details>
<summary>click to reveal</summary>

Greedy by **end time**. Sort intervals by their end value. Walk through, tracking the end of the last kept interval. If the next interval starts before that end, it overlaps — remove it (count++). Otherwise keep it and update the end.

Why end-time and not start-time? Sorting by end greedily picks the interval that frees up the timeline soonest, leaving the most room for later intervals. This is the same trick as "max non-overlapping meetings".
</details>

## My solution

```python
from typing import List

class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        intervals.sort()
        count = 0
        u = intervals[0][1]
        
        for i in range (1,len(intervals)):
            if(intervals[i][0] < u):
                count+=1
                u = min(intervals[i][1],u)
            else:
                u = intervals[i][1]
            
        return count
```

## Test cases

Run after your implementation. All should print `PASS`.

```python
tests = [
    # (intervals, expected_removals, label)
    ([[1,2],[2,3],[3,4],[1,3]],          1, 'classic example from LC'),
    ([[1,2],[1,2],[1,2]],                 2, 'three identical, keep one remove two'),
    ([[1,2],[2,3]],                       0, 'touching at endpoint is NOT overlap'),
    ([[1,2]],                             0, 'single interval'),
    ([[1,100],[11,22],[1,11],[2,12]],     2, 'mixed sizes'),
    ([[-100,-50],[-60,-40],[-30,-10]],    1, 'negative values, two overlap'),
    ([[0,2],[1,3],[2,4],[3,5],[4,6]],     2, 'zigzag overlap chain'),
    ([[1,5],[2,3],[3,4]],                 1, 'one big interval covers two small (remove [1,5])'),
    ([[1,10],[2,3],[4,5],[6,7],[8,9]],    1, 'one wide interval, four narrow inside (remove [1,10])'),
]

sol = Solution()
for intervals, expected, label in tests:
    got = sol.eraseOverlapIntervals([list(x) for x in intervals])
    status = 'PASS' if got == expected else 'FAIL'
    print(f'{status:5}  expected={expected}  got={got}   {label}')
```
